# Data

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.random.normal(5, 2.5, 1000).reshape(-1, 1)
print(X.shape)
X[:10]
```

```python
Y = 3 * X + 2
print(Y.shape)
Y[:10]
```

```python
epsilon = np.random.normal(0, 10, size=X.shape)
Y_rand = Y + epsilon
print(Y_rand.shape)
Y_rand[:10]
```

```python
plt.scatter(X, Y_rand, color="blue")
plt.plot(X, Y, color="lightgreen")
plt.show(block=True)
```

# Regressão Linear: solução analítica

```python
data = X.copy()
data = np.column_stack((np.ones(X.shape[0]), data))
data
```

```python
beta = np.linalg.inv(data.T @ data) @ data.T @ Y
print(beta.shape)
beta
```

```python
# beta0 + beta1 * x
x_min = X.min()
x_max = X.max()

plt.scatter(X, Y_rand, color="blue")
plt.plot(X, Y, color="lightgreen", linewidth=3)

plt.plot(
    [x_min, x_max],
    [beta[0] + beta[1] * x_min, beta[0] + beta[1] * x_max],
    color="red",
    linewidth=1
)

plt.show(block=True)
```
